```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
iris = sns.load_dataset("iris")
```

```python
print(f"Columns: {iris.columns}\n")
print(f"Shape: {iris.shape}")
```

```python
iris.info()
```

```python
print(iris.head(10))
```

```python
iris.describe()
```

```python
iris["species"].value_counts()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data = iris,
    x = 'sepal_length',
    y = 'sepal_width',
    hue = 'species',
    palette = 'Set1'
)
plt.title("Sepal length Vs Sepal Width by Species")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.tight_layout()
plt.show()
```

```python
iris.drop('species', axis=1).hist(bins=20,
                                 figsize = (8, 6),
                                 color = 'steelblue',
                                 edgecolor = 'black')
plt.suptitle('Distribution of All Features', y=1.02)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(data=iris)

plt.title("Box Plot of Iris Features")
plt.show()
```

```python
# Conclusion
'''
- The Iris dataset contains 150 flower samples.
- There are three species: Setosa, Versicolor, and Virginica.
- Petal length and petal width clearly separate the species.
- Box plots help identify spread and possible outliers.
- Histograms show the distribution of feature values.
'''
```
